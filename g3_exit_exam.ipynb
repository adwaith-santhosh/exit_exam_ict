```python
for col in df_num_cols :
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower) |
        (df[col] > upper)
    ]

    print(f"{col}:")
    print(f"Lower limit: {lower}")
    print(f"Upper limit: {upper}")
    print(f"Number of outliers: {len(outliers)}")
    print()
```

```python
for col in df_num_cols :
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[col]=df[col].clip(lower,upper)
```

```python
df = pd.get_dummies(df,columns = df_cat_cols,drop_first = True,dtype = int)
```

df is the cleaned dataset for modeling

```python
y=df['revenue']
x=df.drop(columns=['revenue'])
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler,RobustScaler
```

```python
df_num_cols
```

```python
scal_cols=['employees', 'employees_year', 'revenue_year', 'net_income',
       'net_income_year', 'market_cap', 'legal_units_count',
       'direct_subsidiaries_count', 'max_hierarchy_depth', 'market_cap_day',
       'market_cap_month', 'market_cap_year']
```

```python
std_scaler=StandardScaler()
x_train[scal_cols]=std_scaler.fit_transform(x_train[scal_cols])
x_test[scal_cols]=std_scaler.transform(x_test[scal_cols])
```

```python
x_train
```

before model training the missing values were handled

identified outliers and clipped appropriate columns

categorical columns were encoded with one hot encoding

scaling was done with standard scaler

encoding in this dataset is necessary as there are features with numerous categories

here one hot encoding is done

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
poly = PolynomialFeatures(degree = 2)
x_train_poly = poly.fit_transform(x_train)
x_test_poly = poly.transform(x_test)
lr_model = LinearRegression()
lr_model.fit(x_train_poly,y_train)
y_pred = lr_model.predict(x_test_poly)
y_pred
```

```python
comparison = pd.DataFrame({
    'Actual': y_test.values,
    'Predicted': y_pred
})

comparison.head()
```

```python
from sklearn.metrics import mean_squared_error, r2_score

print("MSE:", mean_squared_error(y_test, y_pred))
print("R2 Score:", r2_score(y_test, y_pred))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
residuals = y_test - y_pred
sns.histplot(residuals,kde = True)
plt.title('Residuals Distribution')
plt.show()
```

```python
from sklearn.ensemble import RandomForestRegressor
rf_model = RandomForestRegressor(n_estimators=100,random_state=42,n_jobs=-1)
rf_model.fit(x_train, y_train)
y_pred_rf = rf_model.predict(x_test)
y_pred_rf
```

```python
print("MSE:", mean_squared_error(y_test, y_pred_rf))
print("R2 Score:", r2_score(y_test, y_pred_rf))
```

```python
from sklearn.ensemble import GradientBoostingRegressor
gb_model = GradientBoostingRegressor( n_estimators=100,learning_rate=0.05,max_depth=3,random_state=42
)
gb_model.fit(x_train, y_train)
y_pred_gb = gb_model.predict(x_test)
y_pred_gb
```

```python
print("MSE:", mean_squared_error(y_test, y_pred_gb))
print("R2 Score:", r2_score(y_test, y_pred_gb))
```

```python
from xgboost import XGBRegressor
xgb_model = XGBRegressor(n_estimators=200,learning_rate=0.05,max_depth=6,subsample=0.8,colsample_bytree=0.8,random_state=42,n_jobs=-1,objective="reg:squarederror")
xgb_model.fit(x_train, y_train)
y_pred_xgb = xgb_model.predict(x_test)
y_pred_xgb
```

```python
print("MSE:", mean_squared_error(y_test, y_pred_xgb))
print("R2 Score:", r2_score(y_test, y_pred_xgb))
```

```python
import numpy as np
import pandas as pd
feature_names = poly.get_feature_names_out(x_train.columns)
coefficients = np.asarray(lr_model.coef_).ravel()
print("Feature names:", len(feature_names))
print("Coefficients:", len(coefficients))
```

```python
feature_importances = pd.DataFrame({
    "feature": feature_names,
    "coefficient": coefficients,
    "absolute_importance": np.abs(coefficients)
}).sort_values(
    by="absolute_importance",
    ascending=False
)

print(feature_importances.head(20))
```

```python
top_features = feature_importances.head(10).sort_values(
    by="absolute_importance",
    ascending=True
)

plt.figure(figsize=(12, 8))

sns.barplot(
    x="absolute_importance",
    y="feature",
    data=top_features,
    color="steelblue"

)

plt.title("Top 10 Polynomial Feature Importances")
plt.xlabel("Absolute coefficient magnitude")
plt.ylabel("Feature")
plt.tight_layout()
plt.show()
```
